# Module 9: Numerical Computation with NumPy
## Array Vectorization & Statistical Aggregation


```python
import numpy as np

prices_pkr = np.array([1500, 2500, 8500, 32000, 1200])
prices_with_tax = prices_pkr * 1.18
print('Original:', prices_pkr)
print('With Tax:', np.round(prices_with_tax, 2))
```

```python
sales_data = np.array([145000, 98000, 210000, 310000, 175000, 320000])
print(f'Mean   : PKR {np.mean(sales_data):,.2f}')
print(f'Median : PKR {np.median(sales_data):,.2f}')
print(f'StdDev : PKR {np.std(sales_data):,.2f}')
```
