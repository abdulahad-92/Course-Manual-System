### 1. Variables & Type Inspection (`type()`)

```python
product_code = "SKU-9942"
unit_price = 1450.75
units_in_stock = 120
is_taxable = True

print("Product Code :", product_code, "| Type:", type(product_code))
print("Unit Price   :", unit_price,   "| Type:", type(unit_price))
print("In Stock     :", units_in_stock, "| Type:", type(units_in_stock))
print("Taxable      :", is_taxable,     "| Type:", type(is_taxable))
```

### 2. Payroll Calculation & Formatted f-Strings (`:,.2f`)

```python
employee_name = "Ali Khan"
hours_worked = 42.5
hourly_rate = 950.0

# Calculate earnings
gross_pay = hours_worked * hourly_rate
tax_withheld = gross_pay * 0.05
net_pay = gross_pay - tax_withheld

print(f"Employee Name : {employee_name}")
print(f"Gross Pay     : PKR {gross_pay:,.2f}")
print(f"Tax (5%)      : PKR {tax_withheld:,.2f}")
print(f"Net Earnings  : PKR {net_pay:,.2f}")
```

### 3. Currency Converter with Precision Formatting

```python
# Constant exchange rates (PKR Base)
USD_RATE = 278.50
EUR_RATE = 302.10

amount_pkr = 150000.0  # PKR 150,000

usd_val = amount_pkr / USD_RATE
eur_val = amount_pkr / EUR_RATE

print(f"PKR {amount_pkr:,.2f} is approximately:")
print(f"  USD : ${usd_val:,.2f}")
print(f"  EUR : €{eur_val:,.2f}")
```
