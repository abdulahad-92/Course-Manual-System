### 1. Commercial Bank Loan Screening (`if / elif / else`)

```python
credit_score = 680
annual_revenue = 650000  # PKR 650,000

print("--- LOAN APPLICATION STATUS ---")

if credit_score >= 700 and annual_revenue >= 500000:
    print("Decision: Loan Approved — Prime Tier Rate (12%)")
elif 600 <= credit_score <= 699 and annual_revenue >= 500000:
    print("Decision: Loan Approved — Standard Tier Rate (15%)")
elif annual_revenue < 500000:
    print("Decision: Loan Denied — Insufficient Annual Revenue")
else:
    print("Decision: Loan Denied — High Credit Risk")
```

### 2. Batch Inventory Revenue Auditor (`for` loop & accumulator)

```python
daily_sales = [120000, 85000, 210000, 95000, 310000, 140000, 90000]
total_weekly_revenue = 0
high_value_days = 0

for day_idx, sale_amount in enumerate(daily_sales, start=1):
    total_weekly_revenue += sale_amount
    if sale_amount >= 200000:
        high_value_days += 1
        print(f"Day {day_idx}: High-Value Transaction Flagged -> PKR {sale_amount:,.2f}")

print("------------------------------------------")
print(f"Total Weekly Revenue : PKR {total_weekly_revenue:,.2f}")
print(f"High-Value Sales Days: {high_value_days}")
```

### 3. ATM PIN Authentication Simulation (`while` loop)

```python
CORRECT_PIN = "1030"
max_attempts = 3
attempts = 0

while attempts < max_attempts:
    user_attempt = "1030" if attempts == 1 else "0000"  # Simulated input for demo
    attempts += 1
    
    if user_attempt == CORRECT_PIN:
        print(f"Attempt {attempts}: PIN Accepted. Access Granted.")
        break
    else:
        remaining = max_attempts - attempts
        print(f"Attempt {attempts}: Incorrect PIN. Attempts remaining: {remaining}")
else:
    print("ACCOUNT LOCKED: Maximum authentication attempts exceeded.")
```
